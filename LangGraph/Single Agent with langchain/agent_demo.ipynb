```python
import os
import certifi
import requests
from dotenv import load_dotenv
```

```python
from langchain_mistralai import ChatMistralAI
from langchain.tools import tool
from langchain_tavily import TavilySearch
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
from langchain.agents import create_agent #react stands for reasoning and action agent
```

```python
os.environ["SSL_CERT_FILE"]=certifi.where() #used to handle some path related issues
load_dotenv()
```

```python
OPENAI_API_KEY=os.getenv("OPENAI_API_KEY")
TAVILY_API_KEY=os.getenv("TAVILY_API_KEY")
MISTRALAPI_KEY=os.getenv("MISTRAL_API_KEY")
```

initializing tavily search tool


```python
search_tool=TavilySearch(max_results=2)
```

```python
res=search_tool.invoke("What is a w16 engine?")
```

```python
print(res['results'])
```

```python
for item in res['results']:
    print(item["content"])
```

Initiallizing LLM

```python
llm=ChatMistralAI(
    model="mistral-large-latest",
    temperature=0,
    api_key=MISTRALAPI_KEY
)
```

```python
res=llm.invoke("hello what is the name of president of india")
```

```python
print(res.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer the question as best as you can, giving examples."),
    ("human", "{input}"),
    MessagesPlaceholder("agent_scratchpad"),
])
```

```python
tools=[search_tool]
```

```python
#create agent
system_prompt = "Answer the question as best as you can, giving examples."
agent=create_agent(model=llm,tools=tools,system_prompt=system_prompt)
```

```python
res=agent.invoke({"messages": [
        {"role": "user", "content": "What is a W16 engine?"}
    ]})
```

```python
print(res["messages"][-1].content)
```

```python
from langchain_classic.agents import AgentExecutor
from langchain_classic.agents import create_tool_calling_agent
from langchain_classic.agents import (
    AgentExecutor,
    create_tool_calling_agent,
    tool,
)
```

```python
agent = create_tool_calling_agent(llm, tools, prompt) #it needs prompt as chat prompt template
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

agent_executor.invoke({"input": "What is the population of capital of india"})
```
