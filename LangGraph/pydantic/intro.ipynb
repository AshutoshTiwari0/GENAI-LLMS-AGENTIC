```python
from pydantic import BaseModel
```

```python
#STEP 1 define pydantic model(class)
class PatientData(BaseModel): #we need to pass BaseModel as parameter 
    name:str 
    age:int
    weight:float

def add_patient_data(patient:PatientData):
    print(f"Name of patient is {patient.name} and age is {patient.age} having weight {patient.weight}")
    print("successfully added")
        
def update_patient_data(patient:PatientData):
    print(f"Name of patient is {patient.name} and age is {patient.age} having weight {patient.weight}")
    print("successfully updated")


#STEP 2 give raw input
patient_information={"name":"Ashutosh",
                     "age":25,
                     "weight":79} # even if we pass "25" we will not get any error



#STEP 3 Pass validated model to functions(for using)
patient=PatientData(**patient_information) # ** unpacks the data

add_patient_data(patient)
update_patient_data(patient)
```

```python
from pydantic import BaseModel
from typing import List,Dict

class PatientData(BaseModel): #we need to pass BaseModel as parameter 
    name:str 
    age:str
    weight:float
    married:bool
    allergies:List[str]
    contact_info:Dict[str,str] #key should be str and value also str

    #schema khud sochna padta hai

def add_patient_data(patient:PatientData):
    print(f"Name of patient is {patient.name} and age is {patient.age} having weight {patient.weight}")
    print("successfully added")
        
def update_patient_data(patient:PatientData):
    print(f"Name of patient is {patient.name} and age is {patient.age} having weight {patient.weight}")
    print("successfully updated")



patient_information={"name":"Ashutosh",
                     "age":"25", #now it expects str 
                     "weight":79.25, # even if we pass "25" we will not get any error
                    "married":False,
                    "allergies":['Dust','Mushrooms'],
                    "contact_info":{'email':'xyz@abc.com',
                                    'phone':"123456789"}
}


patient=PatientData(**patient_information) # ** unpacks the data

add_patient_data(patient)
update_patient_data(patient)
```

field validator

```python


from pydantic import field_validator


class PatientData(BaseModel): #we need to pass BaseModel as parameter 
    name:str 
    email:EmailStr
    
    @field_validator('email') #jis field pe validate krna hai 
    @classmethod
    def email_validator(cls,value):

        valid_domain=['hdfc.com','icici.com']
        # abc@hdfc.com, abc@icici.com are only valid 
        domain_name=value.split('@')[1]

        if domain_name in valid_domain:
            return value
        else:
            raise ValueError("Not a valid domain")
    
def add_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.email)
    print("data added succesfuly")


patient_data={"name":"Ashutosh",
                "email":"ash@icici.com"}

patient1=PatientData(**patient_data)

add_patient_data(patient1)
        
    
```

Model Validator -> used to validate multiple fields at once

```python


from pydantic import model_validator


class PatientData(BaseModel): #we need to pass BaseModel as parameter 
    name:str 
    email:EmailStr
    age:int
    contact_details:Dict[str,str]
    
    @model_validator(mode='after') 
    @classmethod
    def validate_emergency_contact(cls,model):
        if model.age>60 and 'emergency' not in model.contact_details:
            raise ValueError('Patients older than 60 must have emergency contacts')
        return model
    
def add_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.email)
    print("data added succesfuly")


patient_data={"name":"Ashutosh",
                "email":"ash@icici.com",
                "age":98,
                "contact_details":{'name':'Willy Wonka',
                                   'emergency':"Ravi Kishan"}}

patient1=PatientData(**patient_data)

add_patient_data(patient1)
        
    
```

Computed Field-> Given 2 fields find value of another field

```python


from pydantic import computed_field


class PatientData(BaseModel): #we need to pass BaseModel as parameter 
    name:str 
    email:EmailStr
    age:int
    weight:int
    height:int
    contact_details:Dict[str,str]
    
    @computed_field
    @property
    def bmi(self)->float:
        bmi=round(self.weight/(self.height**2),2)
        return bmi


    
def add_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.email)
    print("BMI ",patient.bmi)
    print("data added succesfuly")


patient_data={"name":"Ashutosh",
                "email":"ash@icici.com",
                "age":98,
                "height":58,
                "weight":80,
                "contact_details":{'name':'Willy Wonka',
                                   'emergency':"Ravi Kishan"}}

patient1=PatientData(**patient_data)

add_patient_data(patient1)
        
    
```

Nested Models

```python
from pydantic import BaseModel

class Address(BaseModel):

    city: str
    state: str
    pin: str

class PatientData(BaseModel):

    name: str
    gender: str
    age: int
    address: Address

address_dict = {'city': 'gurgaon', 'state': 'haryana', 'pin': '122001'}

address1 = Address(**address_dict)

patient_dict = {'name': 'Kishor', 'gender': 'male', 'age': 40, 'address': address1}

patient1 = PatientData(**patient_dict)

print(patient1)
print(patient1.name)
print(patient1.address)
print(patient1.address.city)
```
