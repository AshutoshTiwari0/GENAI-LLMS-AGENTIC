```python
import time
```

sychronous code

```python
def fetch_weather():
    print("fetching weather data...")
    time.sleep(4)
    print("data fetched for weather")

def fetch_news():
    print("fetching news data...")
    time.sleep(4)
    print("data fetched for news")

def main():

    start_time=time.time()
    fetch_weather()
    fetch_news()

    end_time=time.time()

    print(f"total time taken is {end_time-start_time} seconds")


main()
```

asychronous code

```python
import asyncio
```

```python
async def fetch_weather():
    print("fetching weather data...")
    await asyncio.sleep(4)
    print("data fetched for weather")

async def fetch_news():
    print("fetching news data...")
    await asyncio.sleep(4)
    print("data fetched for news")

async def main():

    start_time=time.time()
    await asyncio.gather(fetch_weather(),fetch_news())

    end_time=time.time()

    print(f"total time taken is {end_time-start_time} seconds")


await main()
```
